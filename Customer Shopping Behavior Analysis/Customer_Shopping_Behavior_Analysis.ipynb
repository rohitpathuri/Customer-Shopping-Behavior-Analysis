```python
#To check any missing values
df.isnull().sum()
```

```python
# Now we found that we have missing values in Review Rating. so normally as a beginner we will choose median and mean to fill those null values,
# but generally people will choose median over mean that is beacause mean include all the outliers in the result which is not suitable to fiil null values
# but median dosen't include any outliers.
```

```python
# Here, if we just add median it looks like as a normal cleaning process but if you want to fill those null values with some meaningful data then 
# go ahead with this procedure. 
# As we thought to fill with median in review rating, but review rating consist of two categrioes and each category has different ratings. so if just add median 
# as combined values it will differ in rating.  
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# To convert column names into lower case and to replace space(' ') with underscore(_)
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df= df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.columns
```

```python
# To create age_group
labels = ['Young Adult','Adult','Middle-age','senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
#create a colums purchase_frequency_days

frequency_mapping ={
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
